## Step 6: Create Disease Mapping

The trained model predicts encoded disease labels. This step creates a mapping between encoded labels and disease names so that predictions can be displayed in a human-readable format.

```python
# Get all unique diseases from the dataset
disease_names = sorted(df["disease"].unique())

# Create mapping
disease_mapping = {i: disease for i, disease in enumerate(disease_names)}

print("Total Diseases:", len(disease_mapping))
print(disease_mapping)
```

## Step 7: Predict Disease

This function accepts patient information, predicts the disease using the trained Random Forest model, and returns the predicted disease label.

```python
def predict_disease(patient_data):

    patient_df = pd.DataFrame([patient_data])

    prediction = model.predict(patient_df)

    disease = label_encoder.inverse_transform(prediction)[0]

    return disease
```

```python
patient = {
    "fever": 1,
    "cough": 1,
    "fatigue": 1,
    "difficulty_breathing": 0,
    "age": 35,
    "gender": 1,
    "blood_pressure": 120,
    "cholesterol_level": 180,
    "age_scaled": 0.35,
    "bp_scaled": 0.45,
    "chol_scaled": 0.40,
    "risk_level": 1
}

patient_df = pd.DataFrame([patient])

patient_df
```

```python
predicted_disease = model.predict(patient_df)[0]

print("Predicted Disease:", predicted_disease)
```
