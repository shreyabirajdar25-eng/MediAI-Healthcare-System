## Step 9: Disease Distribution Across Age Groups (Bubble Chart)

This visualization represents the relationship between diseases and patient age using bubbles. The bubble size indicates the frequency of cases, making it easier to identify which diseases are common in different age groups.

```python
import matplotlib.pyplot as plt

# Count disease occurrences by age
bubble_data = (
    df.groupby(['disease', 'age'])
      .size()
      .reset_index(name='count')
)

# Take top 15 diseases for better visualization
top_diseases = df['disease'].value_counts().head(15).index
bubble_data = bubble_data[bubble_data['disease'].isin(top_diseases)]

plt.figure(figsize=(14,8))

plt.scatter(
    bubble_data['age'],
    bubble_data['disease'],
    s=bubble_data['count']*40,
    alpha=0.6
)

plt.xlabel("Age")
plt.ylabel("Disease")
plt.title("Disease Distribution Across Age Groups (Bubble Chart)")
plt.grid(alpha=0.3)

plt.show()
```

## Step 10: Symptom Correlation Heatmap

This heatmap visualizes the relationship among different symptoms in the dataset. It helps identify which symptoms frequently occur together, making it easier to understand disease patterns and feature dependencies.

```python
pip install seaborn
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select symptom columns
symptoms = [
    'fever',
    'cough',
    'fatigue',
    'difficulty_breathing'
]

# Convert Yes/No into 1/0
heatmap_df = df[symptoms].replace({'Yes':1, 'No':0})

# Correlation Matrix
corr = heatmap_df.corr()

# Plot Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between Symptoms")
plt.show()
```

## Step 11: Disease Outcome Distribution

This stacked bar chart illustrates the distribution of patient outcomes across different diseases. It helps compare the severity and recovery patterns associated with each disease.

```python
import matplotlib.pyplot as plt

# Count disease vs outcome
outcome_data = pd.crosstab(df['disease'], df['outcome_variable'])

# Top 10 diseases
top10 = df['disease'].value_counts().head(10).index
outcome_data = outcome_data.loc[top10]

# Plot
outcome_data.plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title("Disease vs Outcome Distribution")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.legend(title="Outcome")
plt.tight_layout()

plt.show()
```

## Step 12: Age Distribution by Gender (Violin Plot)

This violin plot visualizes the distribution of patient ages across different genders. It combines a box plot with a density plot, providing insights into the spread, concentration, and variability of age within each gender group.

## Step 13: Pair Plot of Numerical Features

This pair plot visualizes the relationships among numerical features in the healthcare dataset. It helps identify trends, correlations, clusters, and potential outliers while comparing multiple variables simultaneously.

## Step 14: Disease-wise Age Distribution (Box Plot)

This box plot represents the distribution of patient ages for different diseases. It helps identify the median age, variability, and potential outliers, providing insights into how age is associated with various diseases.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Top 10 diseases
top10 = df['disease'].value_counts().head(10).index

plt.figure(figsize=(14,6))

sns.boxplot(
    data=df[df['disease'].isin(top10)],
    x='disease',
    y='age',
    palette='Set3'
)

plt.xticks(rotation=45)
plt.title("Age Distribution Across Top Diseases")
plt.xlabel("Disease")
plt.ylabel("Age")

plt.tight_layout()
plt.show()
```

## Step 15: Gender Distribution Across Diseases

This grouped bar chart compares the number of male and female patients affected by different diseases. It helps identify gender-specific disease prevalence and supports demographic analysis.

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = df['disease'].value_counts().head(10).index

plt.figure(figsize=(14,6))

sns.countplot(
    data=df[df['disease'].isin(top10)],
    x='disease',
    hue='gender'
)

plt.xticks(rotation=45)
plt.title("Gender Distribution Across Diseases")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

# 🎯 Step 28: Relationship Between Age and Cholesterol Level

# 📈 Step 29: Average Age by Disease

This bar chart displays the average age of patients for the most common diseases. It helps identify which diseases are more prevalent among different age groups.

```python
top_avg = (
    df.groupby("disease")["age"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_avg.index,
    y=top_avg.values,
    hue=top_avg.index,
    palette="viridis",
    legend=False
)

plt.xticks(rotation=45)

plt.title("Average Age by Disease")
plt.xlabel("Disease")
plt.ylabel("Average Age")

plt.show()
```
